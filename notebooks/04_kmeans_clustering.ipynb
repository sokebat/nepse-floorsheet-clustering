# Plain K-Means on stocks
**What it does:** picks k random data points as starting centroids, then repeats assign-points-to-nearest-centroid / recompute-centroid-as-mean until nothing changes.

**The problem it has:** the result depends on those random starting points. A bad draw (e.g. two centroids starting close together) can converge to a worse clustering than a good draw would — so the same algorithm on the same data can give different answers depending on luck. `05_kmeans_plusplus.ipynb` covers the fix.

Load env vars and the cleaned data.

```python
import os
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from dotenv import load_dotenv
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

load_dotenv(Path("..") / ".env")

CLEAN_PATH = Path("..") / os.environ["CLEAN_FLOORSHEET_PATH"]

df = pd.read_parquet(CLEAN_PATH)
df.shape
```

Aggregate trades to one row per symbol.

```python
stock_df = df.groupby("symbol").agg(
    total_trades=("transaction", "count"),
    total_turnover=("amount", "sum"),
    total_quantity=("quantity", "sum"),
    avg_rate=("rate", "mean"),
    std_rate=("rate", "std"),
    n_trading_days=("date", "nunique"),
)
stock_df.shape
```

Distinct brokers per symbol (breadth of ownership).

```python
broker_by_symbol = pd.concat([
    df[["symbol", "buyer"]].rename(columns={"buyer": "broker"}),
    df[["symbol", "seller"]].rename(columns={"seller": "broker"}),
])

stock_df["n_brokers"] = broker_by_symbol.groupby("symbol")["broker"].nunique()
```

```python
stock_df["avg_trade_size"] = stock_df["total_turnover"] / stock_df["total_trades"]
stock_df["avg_daily_volume"] = stock_df["total_quantity"] / stock_df["n_trading_days"]

# coefficient of variation instead of raw std - price *levels* differ hugely across stocks
# (a rate of 9 vs. a rate of 850), so raw std would just reflect price level, not volatility
stock_df["price_cv"] = stock_df["std_rate"] / stock_df["avg_rate"]

stock_df.head()
```

Drop symbols with <30 trades — too thin to trust.

```python
MIN_TRADES = 30
before = len(stock_df)

stock_df = stock_df[stock_df["total_trades"] >= MIN_TRADES].copy()

after = len(stock_df)
print(f"dropped {before - after} thinly-traded symbols ({before} -> {after})")
```

Final feature set for clustering.

```python
feature_cols = [
    "total_trades",
    "total_turnover",
    "avg_trade_size",
    "avg_daily_volume",
    "n_brokers",
    "price_cv",
]

stock_df[feature_cols].describe()
```

Log-transform the skewed columns before scaling.

```python
log_cols = ["total_trades", "total_turnover", "avg_trade_size", "avg_daily_volume"]

features = stock_df[feature_cols].copy()
features[log_cols] = np.log1p(features[log_cols])
features.head()
```

Standard-scale so no feature dominates by units.

```python
scaler = StandardScaler()
X = scaler.fit_transform(features)
X.shape
```

Sweep k=2..8, compare inertia (elbow) vs silhouette.

```python
k_range = range(2, 9)
inertias = []
silhouette_scores = []

for k in k_range:
    km = KMeans(n_clusters=k, init="random", n_init=10, random_state=42)
    labels = km.fit_predict(X)
    inertias.append(km.inertia_)
    silhouette_scores.append(silhouette_score(X, labels))

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].plot(list(k_range), inertias, marker="o")
axes[0].set_xlabel("k")
axes[0].set_ylabel("inertia")
axes[0].set_title("Elbow method")

axes[1].plot(list(k_range), silhouette_scores, marker="o", color="darkorange")
axes[1].set_xlabel("k")
axes[1].set_ylabel("silhouette score")
axes[1].set_title("Silhouette score")

fig.tight_layout()
plt.show()
```

```python
best_k = list(k_range)[int(np.argmax(silhouette_scores))]
print(f"best k by silhouette score: {best_k}")
```

Fit the final model with random init (`n_init=10` restarts, kept so the result is still usable).

```python
kmeans = KMeans(n_clusters=best_k, init="random", n_init=10, random_state=42)
stock_df["cluster"] = kmeans.fit_predict(X)

stock_df["cluster"].value_counts().sort_index()
```

PCA to 2D just to visualize the clusters.

```python
pca = PCA(n_components=2, random_state=42)
X_2d = pca.fit_transform(X)

fig, ax = plt.subplots(figsize=(7, 6))
scatter = ax.scatter(
    X_2d[:, 0], X_2d[:, 1], c=stock_df["cluster"], cmap="tab10", s=40, edgecolor="black", alpha=0.8
)
ax.set_xlabel(f"PC1 ({pca.explained_variance_ratio_[0]:.0%} var)")
ax.set_ylabel(f"PC2 ({pca.explained_variance_ratio_[1]:.0%} var)")
ax.set_title("Stocks by cluster (PCA projection)")
legend = ax.legend(*scatter.legend_elements(), title="cluster")
ax.add_artist(legend)
fig.tight_layout()
plt.show()
```

Cluster profile: mean of raw features per cluster.

```python
cluster_profile = stock_df.groupby("cluster")[feature_cols].mean().round(2)
cluster_profile["n_stocks"] = stock_df.groupby("cluster").size()
cluster_profile
```

Top symbols per cluster — sanity check against real tickers.

```python
for cluster_id, group in stock_df.groupby("cluster"):
    top_symbols = group["total_turnover"].sort_values(ascending=False).head(5)
    print(f"cluster {cluster_id} ({len(group)} stocks) top by turnover:")
    print(top_symbols)
    print()
```

Save the result, kept separate from `05`'s output for clarity.

```python
OUT_PATH = Path("..") / os.environ["DATA_PROCESSED_DIR"] / "stock_clusters_kmeans.csv"
stock_df.to_csv(OUT_PATH)
print(f"Wrote {len(stock_df)} stock rows to {OUT_PATH}")
```

## Summary
- 444/516 stocks kept (30+ trades), log-transformed skewed features, scaled, k=2 by silhouette.
- Plain K-Means: random centroid init, `n_init=10` restarts to reduce (not eliminate) the risk of a bad starting draw.
- The problem: results aren't guaranteed consistent run-to-run without those restarts, and restarts just brute-force around the issue rather than fixing it.
- `05_kmeans_plusplus.ipynb` fixes the root cause with smarter centroid seeding.
- Result saved to `stock_clusters_vanilla.csv`.
